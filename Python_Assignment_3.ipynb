## Number Guessing Game

```python
# 1. Setting up the Game

import random
secret_number = random.randint(1, 10)
attempts = 3
```

```python
# 2. setting the prompt for User
while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

# 3. Implementing the guessing logic 

    if guess < 1 or guess > 10:   # If the guess is out of the valid range (1 to 10),
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue
    
    attempts -= 1  # Reduce remaining attempts by 1 for a valid guess

    if guess > secret_number: # If the guess is greater than the secret number,
        print("Too high. Try again.")
    elif guess < secret_number: # If the guess is lower than the secret number,
        print("Too low. Try again.")
    else: # If the guess is correct,
        print("Congratulations! You guessed the correct number.")
        break
else:
    print("Better luck next time!")   
```

## Multiplication Table Generator

```python
# 1. Prompt for user to generate and display the Multiplication Table
number = int(input("Enter the number for which you want the multiplication table: ")) 
for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}") 
```

## BMI Calculator

```python
# 1. Defining The Function

def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi
```

```python
# 2. Prompt for the user for weight and height

weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi_result = calculate_bmi(weight, height)
print(f"Your BMI is: {bmi_result:.2f}")
```
